# Demo of the Library Sandbox

```python
from src.sum_impact_assessment.utils.tools import load_living_labs_from_file, load_measures_from_file, load_kpis_from_file
from src.sum_impact_assessment.utils.kpi_normalizer import KPINormalizer
from src.sum_impact_assessment.models import KPIImpactAnalyzer

measures = load_measures_from_file('data/measures.json')
kpis = load_kpis_from_file('data/kpis.json')
living_labs = load_living_labs_from_file('data/living_labs_data.json', kpi_definitions=kpis)


print("KPIs","-" * 40)
print("Number of kpis : ",len(kpis))

print("Number of living labs : ",len(living_labs))
for lab in living_labs:
    print(f"ID: {lab.id}")
    print(f"Name: {lab.name}")
    print(f"Number of KPIs: {len(lab.kpis)}")
    print(f"Number of Measures: {len(lab.measures)}")
    print("-" * 40)


analyzer = KPIImpactAnalyzer(living_labs, measures)
measures_coeff = analyzer.run_analysis()

print("MEASURES AFTER ANALYSIS","-" * 40)
print("Number of measures : ",len(measures))
for measure in measures_coeff:
    print(f"ID: {measure.id}")
    print(f"Name: {measure.name}")
    print(f"Coeff: {measure.coefficient}")
    print(f"MSQ: {measure.msq}")
    print("-" * 40)

```
